```python
import random
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from collections import Counter
```

```python
angle = 11
n_objects = 1000
n_simulations = 10000

total_angles = []
for _ in range(n_simulations):
    choices = [random.choice([-1, 1]) for _ in range(n_objects)]
    total_angles.append((sum(choices)*angle)%360)

bins = 360
plt.hist(total_angles, bins = bins)
plt.show()
sns.kdeplot(total_angles)
plt.xlim([0,360])
```
